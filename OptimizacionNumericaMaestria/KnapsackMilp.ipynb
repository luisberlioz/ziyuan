# Problema de la Mochila (0-1)
El Problema de la Mochila (0-1) puede formularse como un Programa Lineal Entero Mixto (Mixed-Integer Linear Program, MILP) en SageMath, utilizando variables de decisión binarias.

Supongamos que tenemos:

$𝑛$: número de objetos.

$v_i$ : valor (beneficio) del objeto  $𝑖$.

$𝑤_𝑖$: : peso del objeto  $𝑖$.

$W$: capacidad máxima de la mochila.

$𝑥_𝑖 \in \{0, 1\}$ indica si el objeto $𝑖$ es seleccionado.

## Formulación matemática

El objetivo es maximizar el valor total de los objetos seleccionados:

$$\max \sum_{i=1}^n v_i x_i$$

Sujeto a la restricción de capacidad:

$$\sum_{i=1}^n w_i x_i \leq W$$

Y las restricciones de integralidad: $𝑥_𝑖 \in \{0, 1\}$ 

```python
n = 5

# Values (profits)
v = [10, 40, 30, 50, 35]

# Weights
w = [5, 4, 6, 3, 2]

# Knapsack capacity
W = 10

KP = MixedIntegerLinearProgram(maximization=True)
x = KP.new_variable(binary=True)

KP.set_objective(
    sum(v[i] * x[i] for i in range(n))
)

KP.add_constraint(
    sum(w[i] * x[i] for i in range(n)) <= W
)

KP.solve()

solution = KP.get_values(x)
selected = [i for i in range(n) if solution[i] > 0.5]

print("Optimal value:", KP.get_objective_value())
print("Selected items:", selected)
print("Total weight:", sum(w[i] for i in selected))
```
